```python
import pandas as pd

# Caminho absoluto para o CSV para evitar problemas de diretório
csv_path = r"c:\Users\milto\OneDrive\Documentos\SENAI CIMATEC\Machine Learning\curso-ml-cimatec\olist\data\raw\olist_orders_dataset.csv"

# Leitura do dataset
orders_df = pd.read_csv(csv_path)

# Informações gerais da estrutura
print(f"Forma do dataset: {orders_df.shape}")
print("\nColunas:")
print(orders_df.columns.tolist())
print("\nTipos de dados:")
print(orders_df.dtypes)
print("\nPrimeiras linhas:")
display(orders_df.head())
```

```python
# Análise detalhada do dataset de pedidos

print("=" * 80)
print("ANÁLISE DETALHADA: olist_orders_dataset.csv")
print("=" * 80)

# 1. Verificar valores faltantes
print("\n1. VALORES FALTANTES (NaN):")
print(orders_df.isnull().sum())
print("\nPercentual de dados faltantes por coluna:")
print((orders_df.isnull().sum() / len(orders_df) * 100).round(2))

# 2. Status dos pedidos
print("\n2. DISTRIBUIÇÃO DE STATUS DE PEDIDOS:")
print(orders_df["order_status"].value_counts())
print(f"\nTotal de pedidos: {len(orders_df)}")

# 3. Conversão de timestamps para datetime e análise de datas
print("\n3. ANÁLISE TEMPORAL:")
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
]

for col in date_columns:
    orders_df[col] = pd.to_datetime(orders_df[col], errors="coerce")

print(f"Período de pedidos (primeira compra): {orders_df['order_purchase_timestamp'].min()}")
print(f"Período de pedidos (última compra): {orders_df['order_purchase_timestamp'].max()}")

# 4. Clientes únicos
print("\n4. INFORMAÇÕES DE CLIENTES:")
print(f"Total de clientes únicos: {orders_df['customer_id'].nunique()}")
print(f"Pedidos por cliente (média): {len(orders_df) / orders_df['customer_id'].nunique():.2f}")
print(f"Cliente com mais pedidos: {orders_df['customer_id'].value_counts().max()} pedidos")

# 5. Estatísticas de delays
print("\n5. ANÁLISE DE ENTREGA:")
orders_df["delivery_delay_days"] = (
    (orders_df["order_delivered_customer_date"] - orders_df["order_estimated_delivery_date"]).dt.days
)

print(f"Pedidos entregues no prazo: {(orders_df['delivery_delay_days'] <= 0).sum()}")
print(f"Pedidos atrasados: {(orders_df['delivery_delay_days'] > 0).sum()}")
print(f"Atraso médio (dias): {orders_df['delivery_delay_days'].mean():.2f}")
print(f"Atraso máximo (dias): {orders_df['delivery_delay_days'].max()}")

print("\n6. RESUMO ESTATÍSTICO DO DATASET:")
display(orders_df.info())
```
